```python
import pandas as pd

cleaned_df = pd.read_csv("cleaned_student_data.csv")

print("Dataset loaded successfully!")
print(cleaned_df.shape)
```

```python
# Categorical Data Analysis

print("Gender Distribution:")
print(cleaned_df["gender"].value_counts())

print("\nDepartment Distribution:")
print(cleaned_df["department"].value_counts())

print("\nCity Distribution:")
print(cleaned_df["city"].value_counts())
```

```python
# Numerical Variable Distributions

import matplotlib.pyplot as plt

numeric_cols = [
    "age",
    "attendance_pct",
    "study_hours_per_week",
    "previous_score",
    "final_score"
]

cleaned_df[numeric_cols].hist(
    figsize=(12, 8),
    bins=10
)

plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
# Correlation Analysis

import seaborn as sns
import matplotlib.pyplot as plt

numeric_cols = [
    "age",
    "attendance_pct",
    "study_hours_per_week",
    "previous_score",
    "final_score"
]

correlation = cleaned_df[numeric_cols].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Study Hours vs Final Score

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=cleaned_df,
    x="study_hours_per_week",
    y="final_score",
    s=80
)

plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours per Week")
plt.ylabel("Final Score")
plt.grid(True)
plt.show()
```

```python
# Further Exploratory Data Analysis

print("Average Final Score by Gender:")
print(cleaned_df.groupby("gender")["final_score"].mean())

print("\nAverage Final Score by Department:")
print(cleaned_df.groupby("department")["final_score"].mean())

print("\nAverage Final Score by City:")
print(cleaned_df.groupby("city")["final_score"].mean())


# Attendance vs Final Score
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=cleaned_df,
    x="attendance_pct",
    y="final_score"
)
plt.title("Attendance vs Final Score")
plt.xlabel("Attendance Percentage")
plt.ylabel("Final Score")
plt.grid(True)
plt.show()


# Previous Score vs Final Score
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=cleaned_df,
    x="previous_score",
    y="final_score"
)
plt.title("Previous Score vs Final Score")
plt.xlabel("Previous Score")
plt.ylabel("Final Score")
plt.grid(True)
plt.show()


# Final Score by Department
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=cleaned_df,
    x="department",
    y="final_score"
)
plt.title("Final Score by Department")
plt.xlabel("Department")
plt.ylabel("Final Score")
plt.show()
```

```python
# Key EDA Insights

print("=== KEY EDA INSIGHTS ===")

print("\n1. Highest Final Score:")
print(cleaned_df.loc[cleaned_df["final_score"].idxmax()])

print("\n2. Lowest Final Score:")
print(cleaned_df.loc[cleaned_df["final_score"].idxmin()])

print("\n3. Average Final Score:")
print(cleaned_df["final_score"].mean())

print("\n4. Average Attendance:")
print(cleaned_df["attendance_pct"].mean())

print("\n5. Average Study Hours per Week:")
print(cleaned_df["study_hours_per_week"].mean())

print("\n6. Correlation with Final Score:")
print(
    cleaned_df[
        ["age", "attendance_pct", "study_hours_per_week",
         "previous_score", "final_score"]
    ].corr()["final_score"].sort_values(ascending=False)
)
```

```python
# Final EDA Analysis

print("=== FINAL EDA ANALYSIS ===")

# Average scores by gender
print("\nAverage Final Score by Gender:")
print(cleaned_df.groupby("gender")["final_score"].mean())

# Average scores by department
print("\nAverage Final Score by Department:")
print(cleaned_df.groupby("department")["final_score"].mean())

# Average scores by city
print("\nAverage Final Score by City:")
print(cleaned_df.groupby("city")["final_score"].mean())

# Correlation with final score
print("\nCorrelation with Final Score:")
correlation = cleaned_df[
    ["age", "attendance_pct", "study_hours_per_week",
     "previous_score", "final_score"]
].corr()["final_score"].sort_values(ascending=False)

print(correlation)
```

```python
# Summary: Average Final Score by Department

department_avg = cleaned_df.groupby("department")["final_score"].mean()

plt.figure(figsize=(8, 5))
department_avg.plot(kind="bar")

plt.title("Average Final Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Final Score")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## Final Analysis & Conclusion

The cleaned student dataset contains 25 records and 10 columns after removing duplicate records, handling missing values, correcting inconsistent categorical values, and validating numerical ranges.

The analysis shows a strong relationship between academic performance and the following factors:

- **Study hours per week** have the strongest correlation with final score (0.969).
- **Attendance percentage** also has a very strong positive correlation with final score (0.949).
- **Previous score** has a strong positive correlation with final score (0.937).

The department-wise analysis shows that **ECE has the highest average final score (84.50)**, followed closely by **CSE (84.45)**. **ME has the lowest average final score (66.17)**.

The visualizations support these findings. Students who study more hours, maintain higher attendance, and have stronger previous academic performance generally achieve higher final scores.

Overall, the dataset is now cleaned, validated, standardized, analyzed, and exported in a format suitable for further data-science work.
